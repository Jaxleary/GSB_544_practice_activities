```python
def sing_verse(num):
    song = (
        str(num) + " bottles of beer on the wall \n"
        + str(num) + " bottles of beer \n"
        + " take one down, pass it around, \n"
        + str(num - 1) + " bottles of beer on the wall \n"
    )

    return song


song = []

for i in range(100, 97, -1):
    song.append(sing_verse(i))

print(song)
```

    Write a function `sqrt_pos_unvec()` that takes in a single value as an argument, and returns the square root if the value is positive. Then, write a for loop that uses this function to construct a new vector where the positive values are square rooted.

```python
import numpy as np
```

```python
def sqrt_pos_unvec(x):
    if x > 0:
        return np.sqrt(x)
    else:
        return x

nums = np.array(range(-2,12,2))
results = []
for num in nums:
    results.append(sqrt_pos_unvec(num))

print(results)
```

```python
def sqrt_pos_vec(list):
    list_pos = list > 0
    list[list_pos] = np.sqrt(list[list_pos])
    return list
    
sqrt_pos_vec(nums)
```

4.  Complete the Check-in before Section 7.4.
    - a. Write a `sing_verse_3()` function that also lets us change the container (e.g. bottle, can, …) at each step of the loop.
    - b. Use `map()` to sing a few verses.
    - c. What happens if you supply three different drinks, but only two different types of containers? **The loop stops when it hits the end of the containers list. Even though the `nums` range down to 97, the song stops at 98**
    

```python
def sing_verse_3(num, container, drink):
    song = str(num) + f" {container} of " + drink + " on the wall \n"
    song += str(num) + f" {container} of " + drink + "\n"
    song += " take one down, pass it around, \n"
    song += str(num-1) + f" {container} of " + drink + " on the wall \n"

    return song

nums = range(100, 96, -1)
drinks = ["beer", "milk", "lemonade"]
containers = ["cans", "bottles"]
song = map(sing_verse_3, nums, containers, drinks)
print("".join(list(song)))
```

6. Load the `penguins` dataset from the `palmerpenguins` library, as well as any other libraries you need. Then write a function that takes in information about a penguin, and returns one of the following definitions:

*   "Big Mouth Billy": Male penguins with bill length times bill depth greater than 800.
*   "Dainty Daisy": Female penguins with flipper length less than 5% of body mass.
*   "Average Adelie:" Any Adelie penguins that do not fall into either category.
*   "Other": Any penguins that do not fall into any of the categories.



```python
from palmerpenguins import load_penguins
penguins = load_penguins()
```

```python
def classify_penguin(species, sex, bill_length_mm, bill_depth_mm,
                     flipper_length_mm, body_mass_g):
    
    if sex == "male" and bill_length_mm * bill_depth_mm > 800:
        return "Big Mouth Billy"
    
    elif sex == "female" and flipper_length_mm < 0.05 * body_mass_g:
        return "Dainty Daisy"
    
    elif species == "Adelie":
        return "Average Adelie"
    
    else:
        return "Other"


classify_penguin(species="Adelie",
    sex="male",
    bill_length_mm=45,
    bill_depth_mm=20,
    flipper_length_mm=190,
    body_mass_g=4000
)
```

7. Use an iterable function to create a new variable in the data frame called `category_name` that adds these labels.

```python
penguins["category_name"] = penguins.apply(
    lambda row: classify_penguin(
        row["species"],
        row["sex"],
        row["bill_length_mm"],
        row["bill_depth_mm"],
        row["flipper_length_mm"],
        row["body_mass_g"]
    ),
    axis=1
)

penguins[["species", "sex", "category_name"]].head()
```
